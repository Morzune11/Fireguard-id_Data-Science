# FIREGUARD ID — 05: Deep Learning Modeling (Deep Embedded Clustering / DEC)
Implementasi PyTorch Autoencoder + Student's t-distribution clustering refinement (Xie et al., 2016) untuk mempelajari representasi laten fitur kerawanan Karhutla.

```python
import sys
sys.path.append('../src')
from data_loader import load_splits
from modeling import train_dec, predict_dec
from evaluation import evaluate_clustering

train_df, val_df, _ = load_splits('../data/split')
scaled_cols = [c for c in train_df.columns if c.endswith('_scaled')]
X_train = train_df[scaled_cols].values
X_val = val_df[scaled_cols].values

dec_model, dec_train_lbl, dec_train_z = train_dec(
    X_train, n_clusters=4, latent_dim=4, pretrain_epochs=120, refine_epochs=80, random_state=42
)
dec_val_lbl, dec_val_z, _ = predict_dec(dec_model, X_val)

print('DEC Train Metrics (Latent):', evaluate_clustering(dec_train_z, dec_train_lbl))
print('DEC Val Metrics   (Latent):', evaluate_clustering(dec_val_z, dec_val_lbl))
```
