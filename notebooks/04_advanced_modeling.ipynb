# FIREGUARD ID — 04: Advanced ML Modeling (HDBSCAN)
Density-based spatial clustering menggunakan `sklearn.cluster.HDBSCAN` untuk mendeteksi klaster berdensitas tidak seragam serta memisahkan *noise*.

```python
import sys
sys.path.append('../src')
import pandas as pd
from data_loader import load_splits
from modeling import train_hdbscan, predict_hdbscan_nearest
from evaluation import evaluate_clustering

train_df, val_df, _ = load_splits('../data/split')
scaled_cols = [c for c in train_df.columns if c.endswith('_scaled')]
X_train = train_df[scaled_cols].values
X_val = val_df[scaled_cols].values

configs = [(10, 3), (10, 5), (15, 5), (15, 8), (20, 5), (25, 5), (30, 10)]
records = []
for mcs, ms in configs:
    _, lbl = train_hdbscan(X_train, min_cluster_size=mcs, min_samples=ms)
    records.append({'min_cluster_size': mcs, 'min_samples': ms, **evaluate_clustering(X_train, lbl)})

pd.DataFrame(records)
```

```python
hdb_model, hdb_train_lbl = train_hdbscan(X_train, min_cluster_size=15, min_samples=5)
hdb_val_lbl = predict_hdbscan_nearest(hdb_model, X_val)
print('HDBSCAN Train Metrics:', evaluate_clustering(X_train, hdb_train_lbl))
print('HDBSCAN Val Metrics  :', evaluate_clustering(X_val, hdb_val_lbl))
```
