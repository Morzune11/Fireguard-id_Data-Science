## 4.1 & 4.2 Informasi Dataset dan Deskripsi Fitur

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# 1. Muat dataset raw titik api hasil clipping Riau
df_hotspots = pd.read_csv('../riau_viirs_hotspots_2021_2025.csv')
print('Dimensi Dataset Hotspot:', df_hotspots.shape)
df_hotspots.info()
```

```python
# 2. Muat dataset agregasi Grid 0.1° (~11 km x 11 km)
df_grid = pd.read_csv('../riau_vulnerability_grid_01deg.csv')
print('Dimensi Dataset Grid:', df_grid.shape)
df_grid.head()
```

## 4.3 Analisis Kondisi Data (Quality Assessment)
Memeriksa missing values, duplikasi, anomali, dan ketidakseimbangan kelas.

```python
print('Missing Values pada Dataset Hotspots:')
print(df_hotspots.isnull().sum())

print('\nMissing Values pada Dataset Grid:')
print(df_grid.isnull().sum())

print('\nDuplikat Hotspots:', df_hotspots.duplicated().sum())
print('Duplikat Grid:', df_grid.duplicated().sum())
```

## 4.4 Exploratory Data Analysis (EDA) & Visualisasi
Tiga visualisasi utama untuk menemukan pola temporal dan spasial kerawanan karhutla.

```python
# EDA 1: Tren Titik Api Tahunan (2021 - 2025)
df_hotspots['acq_date'] = pd.to_datetime(df_hotspots['acq_date'])
df_hotspots['year'] = df_hotspots['acq_date'].dt.year
yearly = df_hotspots['year'].value_counts().sort_index()

plt.figure(figsize=(8, 4))
bars = plt.bar(yearly.index.astype(str), yearly.values, color='#e67e22', edgecolor='#d35400', width=0.5)
plt.title('Tren Frekuensi Hotspot Tahunan di Provinsi Riau', fontsize=12, fontweight='bold')
plt.xlabel('Tahun')
plt.ylabel('Jumlah Hotspot')
for b in bars:
    plt.text(b.get_x() + b.get_width()/2, b.get_height() + 50, f'{int(b.get_height()):,}', ha='center', fontweight='bold')
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
```

```python
# EDA 2: Siklus Musiman Bulanan (Musim Kemarau vs Hujan)
df_hotspots['month'] = df_hotspots['acq_date'].dt.month
monthly = df_hotspots['month'].value_counts().sort_index()
month_names = ['Jan', 'Feb', 'Mar', 'Apr', 'Mei', 'Jun', 'Jul', 'Agu', 'Sep', 'Okt', 'Nov', 'Des']

plt.figure(figsize=(9, 4))
plt.plot(month_names, [monthly.get(m, 0) for m in range(1, 13)], marker='o', color='#c0392b', lw=2.5)
plt.title('Pola Siklus Musiman Titik Api di Riau (Puncak Kemarau)', fontsize=12, fontweight='bold')
plt.xlabel('Bulan')
plt.ylabel('Total Titik Api')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

```python
# EDA 3: Distribusi Kepadatan Titik Api per Grid Spasial
plt.figure(figsize=(8, 4))
plt.hist(df_grid['hotspot_count'], bins=25, color='#2980b9', edgecolor='black', log=True)
plt.title('Distribusi Titik Api per Unit Grid 0.1° (Log Scale)', fontsize=12, fontweight='bold')
plt.xlabel('Jumlah Hotspot')
plt.ylabel('Banyaknya Grid (Skala Log)')
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
```
